### 셀 01 · 문항 ID와 정답 문단 매핑

질문 자료에 적힌 첫 기대 ID로 원문을 고릅니다.

Q01에서 ‘어디에’라는 질문의 필수 답은 과제함입니다. ‘제출 뒤 파일 이름과 등록 시간을 확인하세요’는 추가 안내로 기록하세요. 팀에서 절차 전체를 필수로 채점하려면 질문도 ‘어디에 내고 무엇을 확인하나요?’로 바꿉니다.


```python
paragraphs = [{'paragraph_id': 'p1',
  'text': '정규 수업은 평일 오전 9시에 시작합니다. 일정 변경은 과정 게시판에서 확인합니다.'},
 {'paragraph_id': 'p2',
  'text': '과제는 학습 관리 페이지의 과제함에 제출합니다. 제출 뒤 파일 이름과 등록 시간을 확인합니다.'},
 {'paragraph_id': 'p3',
  'text': '제출 오류가 발생하면 오류 화면과 발생 시간을 기록하고 담당 강사에게 문의합니다. 비밀키가 보이는 화면은 공유하지 '
          '않습니다.'}]
question = {'id': 'Q01',
 'split': 'development',
 'question': '과제는 어디에 내나요?',
 'answerable': True,
 'expected_paragraph_ids': ['p2'],
 'expected_points': ['과제함', '파일 이름과 등록 시간 확인']}
by_paragraph = {row['paragraph_id']: row['text'] for row in paragraphs}
print(by_paragraph[question['expected_paragraph_ids'][0]])
```

제출 위치와 제출 후 확인 절차가 한 문단에 함께 나옵니다. 검색기가 p3의 “제출 오류”를 골랐다면 단어는 비슷해도 질문의 답은 빠집니다.
