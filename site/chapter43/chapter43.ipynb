### 셀 01 · 수용 기준에서 예외 질문 확인하기

제공 기획서에 적힌 A2를 읽어 보세요. 팀이 점수 조회를 구현하지 않아도 사용자가 볼 답은 정의할 수 있습니다.

A1 · 정상 질문은 ‘과제는 어디에 내나요?’입니다. 운영 안내의 제출 절차 문단을 읽고 ‘학습 관리 페이지의 과제함’이라고 안내하면 됩니다. 이제 A2에서는 점수를 묻는 질문에 어떤 답을 멈춰야 하는지 정해 보세요.

```python
criterion = {'id': 'A2',
 'input': '내 점수는 몇 점인가요?',
 'expected': '문서로 확인할 수 없음을 알리며 점수를 만들지 않는다.',
 'evidence': '답변과 검색 결과 기록'}
print(criterion['input'], criterion['expected'])
```

“내 점수는 몇 점인가요?”에 문서로 확인할 수 없다고 답합니다. 이 조건이면 팀은 개인 성적 데이터를 수집할 이유가 없습니다.
