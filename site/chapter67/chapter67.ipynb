## 실행 시나리오로 평가 분모 산정

67장의 제공 평가 기록 26행에는 선택하지 않은 기억·MCP 기능도 들어 있습니다. 같은 승인 건의 대기와 재개가 다른 행에 기록되기도 합니다. 행 수는 서로 독립적인 사용자 질문 수가 아닙니다.


### 실패한 실행과 실행하지 않은 항목 나누기



실패했더라도 실제로 실행한 사례는 분모에 남습니다. 평가 행을 고르는 조건을 따라가세요.


### 셀 03 · 미실행 시나리오의 분모 제외

제공 기록에서 발췌한 두 행입니다. 실제 로컬 실행을 표시한 행만 골라 봅니다.

정상 질문 4행은 답과 근거를 확인하는 데 사용합니다. 진단 20행은 실패 위치를 찾는 기록이고 미채택 2행은 후보를 쓰지 않은 이유를 남깁니다. 세 묶음을 합친 26행을 모두 정상 정답 문항으로 세지 마세요.


````python
entries = [{"case_id": "E67-search-error", "execution_status": "executed_local", "failure_stage": "retrieval"},
           {"case_id": "E67-memory-unselected", "execution_status": "not_selected"}]
observed = [row for row in entries if row["execution_status"] == "executed_local"]
print(observed[0]["case_id"], observed[0]["failure_stage"])
````

남은 한 건은 검색 시간 초과를 확인한 행입니다. 선택하지 않은 기억 기능을 실패 또는 성공으로 세지 않습니다. 실제 평가표에서는 검색, 도구, 기억, 응답 중 관찰한 위치에만 원인을 적습니다.


### 평가에 실제로 들어가는 기록

<p>observed에는 execution_status가 executed_local인 행만 남습니다. 검색이 실패한 E67-search-error는 실행했으므로 포함하고, E67-memory-unselected는 실행하지 않아 제외합니다.</p>


<h3>답변 오류와 실행 오류는 어떻게 다른가?</h3>

<p>첫 셀은 답이 만들어졌어도 원문과 제출 장소가 어긋날 수 있음을 보여줍니다. 마지막 셀에서 E67-search-error는 실제로 실행했지만 retrieval에서 실패한 사례입니다. 답변이 틀린 경우, 검색에서 실패한 경우, 아직 실행하지 않은 경우를 나누어 평가하세요.</p>


### 실패가 일어난 단계를 보존하기

<p>선택한 행의 failure_stage는 retrieval입니다. 단순히 답변이 없었다고만 적으면 검색 단계에서 실패한 사례와 평가에서 제외한 사례를 나눌 수 없습니다.</p>
