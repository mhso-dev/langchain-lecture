## 인수 문서에 재현 입력과 복구 한계 적기

113장의 복구 기록을 읽고 인수자가 첫 실행에 사용할 질문·예상 답·위치를 한 문서로 묶습니다. 아직 확인하지 못한 요청도 같은 문서에 적습니다.

이 프로젝트의 노트북은 한 폴더에서 열어 주세요. 첫 셀은 `team-project/final/chapter113.json`을 먼저 찾습니다. 팀 파일이 없으면 [이 장의 시작 예시](chapter114-start.json)를 노트북 옆에 내려받으세요. 출력 첫 줄에서 어느 파일을 읽었는지 확인할 수 있습니다.

```python
from pathlib import Path
import json
import re

here = Path.cwd()
project_root = here.parent if re.fullmatch(r"chapter[0-9]+", here.name) else here
team_file = project_root / "team-project/final/chapter113.json"
sample_file = here / "chapter114-start.json"
input_file = team_file if team_file.is_file() else sample_file
if not input_file.is_file():
    raise FileNotFoundError("이 장의 시작 예시를 내려받아 노트북 옆에 두세요.")
incoming = json.loads(input_file.read_text(encoding="utf-8"))
print("팀 기록" if input_file == team_file else "교재 예시", input_file.name)
print(incoming['restore'])
```

두 번째 줄은 읽은 파일의 `restore` 값입니다.

### 셀 01 · 인수 문항과 기대 근거

규정 A의 원문이 담긴 자료를 다시 읽어 인수자가 볼 첫 출력과 위치를 맞춥니다.

이 셀을 실행한 뒤 첫 답과 출처가 위 인수 예시와 일치하는지 확인하세요.

```python
question = '장비 A의 점검 전 확인 사항은?'
rule = {'id': 'RULE-A',
 'equipment': '장비 A',
 'text': '점검 전 전원 표시등을 확인합니다.',
 'source': '가상 장비실 규정 A',
 'page': 1,
 'section': '2.1',
 'revised_on': '2026-11-01',
 'version': 'rules-v1',
 'groups': ['public', 'admin']}
print(question)
print(rule["text"])
print(f'{rule["id"]} · {rule["source"]} {rule["page"]}쪽 {rule["section"]}절 · {rule["revised_on"]}')
```

질문 아래에 전원 표시등, RULE-A, 1쪽 2.1절, 2026-11-01이 보입니다. 인수자는 자기 화면에서 이 값이 다른지 보고 차이가 난 위치부터 알려 줄 수 있습니다.

인계받은 사람이 장비 A의 질문에서 RULE-A 2.1절을 찾았다면 문서를 따라 기본 경로를 재현한 것입니다. 장비 B로 입력을 바꿨을 때는 답변뿐 아니라 근거 위치도 RULE-B 2.2절로 바뀌어야 합니다. 예시 문장을 복사한 구현을 구별할 수 있는 검사입니다.

## 팀 사례에 적용하기 · 인수 문서에 재현 입력과 복구 한계 적기

문서만 받고도 질문을 넣고 어느 절을 열어야 하는지 알 수 있습니다. 공개 자료에 후보가 없으면 source가 비므로 실제 원문을 보완한 뒤 인수하세요.

```python
output_file = project_root / "team-project/final/chapter114.json"
output_file.parent.mkdir(parents=True, exist_ok=True)
with output_file.open("a+", encoding="utf-8") as handle:
    handle.seek(0)
    previous_text = handle.read()
    if previous_text and json.loads(previous_text) != new:
        raise FileExistsError("같은 장의 다른 결과가 있습니다. 기존 JSON을 다른 이름으로 보관한 뒤 다시 실행하세요.")
    if not previous_text:
        json.dump(new, handle, ensure_ascii=False, indent=2)
        handle.write("\n")
print(output_file.relative_to(project_root).as_posix())
```
