## 녹음 길이를 초로 나타내기

화면에는 2분 3초라고 쓰여 있지만 편집 프로그램에는 길이를 초 단위로 넣어야 합니다. 숫자처럼 보이는 두 값이 실제로도 계산할 수 있는 값인지부터 알아봅니다.

### 값에 이름 붙이기

변수는 값을 다시 쓸 때 부르는 이름입니다. 등호 오른쪽을 먼저 계산한 뒤 왼쪽 이름에 그 값을 붙입니다. print()는 괄호 안의 값을 출력합니다.

```python
minutes = 2
seconds = 3
total = minutes * 60 + seconds  # 곱셈을 먼저 하므로 분만 60배가 됩니다.
print(total)
```

123은 120초와 3초를 더한 값입니다. 첫 줄의 2를 3으로 바꾸고 이 셀을 다시 실행하세요. 183이 나오면 1분 늘었을 때 몇 초가 늘었는지 설명할 수 있습니다.

### 같은 기호로 글자 두 개 더하기

따옴표로 감싼 값은 문자열입니다. 숫자의 덧셈에서 쓴 +를 문자열 사이에 쓰면 어떻게 될까요?

```python
minutes_text = "02"
print(minutes_text + "3")  # 두 문자열을 뒤이어 붙입니다.
```

023이 보입니다. 첫 글자 0도 그대로 남았습니다. type(minutes_text)를 print() 안에 넣으면 str이 보입니다. str은 문자열의 타입 이름이고, 정수 2의 타입 이름은 int입니다.

### 입력을 계산하기 전에 숫자로 바꾸기

int()는 "02"처럼 정수로 읽을 수 있는 문자열을 정수로 바꿉니다. 변환 결과에 60을 곱해 봅시다.

```python
minutes_text = "02"
minutes = int(minutes_text)  # 원래 글자는 남겨 두고 계산할 값을 새 이름에 담습니다.
print(minutes * 60)
```

120초입니다. 첫 줄을 "2분"으로 바꾸면 int()가 ValueError를 냅니다. <mark>문자열에 단위 글자까지 있어서 정수로 읽을 수 없습니다.</mark> 오류를 본 뒤 입력을 되돌리세요. 숫자로 바꾸지 않고 "02"에 60을 곱하면 같은 두 글자가 60번 이어집니다.

### 예상한 값과 같은지 묻기

==는 양쪽 값이 같으면 True, 다르면 False를 돌려줍니다. 계산 결과를 손으로 구한 값과 비교할 때 씁니다.

```python
total = 2 * 60 + 3
print(total == 123)  # =는 값을 붙이고, ==는 두 값을 비교합니다.
```

True가 나왔습니다. 오른쪽만 "123"으로 바꾸면 False가 됩니다. 따옴표는 출력에서 보이지 않을 수 있으므로 True가 나오지 않으면 값과 타입을 함께 살펴보세요.

## 녹음 표식을 목록으로 보관하기

녹음할 때 붙인 색상 표식은 같은 색이 여러 번 나올 수 있습니다. 순서까지 담은 원본은 보존하고 사본에서 표식 한 개를 고치려고 합니다. 먼저 위치로 값을 찾는 방법을 연습하세요.

### 문자열의 일부분 읽기

파이썬에서 첫 글자의 위치 번호는 0입니다. 대괄호 안에 시작과 끝 위치를 콜론으로 나눠 쓰면 그 범위를 잘라 읽습니다. 끝 위치의 글자는 포함하지 않습니다.

```python
label = "REC-07"
print(label[4:6])  # 4번과 5번 위치의 글자를 꺼냅니다.
```

07에는 앞자리 0이 남습니다. 한 글자만 읽는 label[4]의 값은 "0"입니다. 표식 번호처럼 앞자리 0도 뜻이 있는 값은 문자열로 두면 모양을 보존할 수 있습니다.

### 리스트의 한 칸 바꾸기

리스트는 대괄호에 여러 값을 순서대로 담습니다. 같은 값도 여러 칸을 차지할 수 있습니다. 리스트 이름 뒤에 위치 번호를 쓰면 그 칸을 읽거나 바꿀 수 있습니다.

```python
colors = ["파랑", "빨강", "파랑"]
colors[1] = "초록"  # 두 번째 표식만 고칩니다.
print(colors)
```

출력 맨 앞과 맨 끝의 파랑은 남습니다. 세 칸짜리 리스트에서 colors[3]을 읽으면 IndexError가 납니다. 위치 번호는 0, 1, 2까지입니다.

### 목록에 두 번째 이름을 붙이면 어떻게 될까요?

같은 녹음 목록을 다른 이름으로 편집할 때 원본도 바뀌는지 살펴봅니다. append는 리스트 끝에 값을 하나 더하는 메서드입니다. 점 앞의 리스트가 바뀝니다.

```python
colors = ["파랑", "빨강"]
editor = colors
editor.append("노랑")  # editor가 가리키는 목록의 끝에 추가합니다.
print(colors)
```

colors로 읽어도 노랑이 보입니다. editor = colors를 실행했을 때 같은 리스트에 이름이 두 개 생겼기 때문입니다. colors.copy()로 새 리스트를 만들면 사본에서 칸을 바꿔도 이 원본의 칸은 그대로입니다.

### 색의 종류만 세기

set은 값마다 하나씩만 보관하는 집합을 만듭니다. len은 항목 수를 셉니다. 두 연산을 함께 쓰면 순서를 정할 필요 없이 색의 종류를 셀 수 있습니다.

```python
colors = ["파랑", "빨강", "파랑"]
unique_colors = set(colors)
print(len(unique_colors))  # 표식 3개 가운데 서로 다른 색의 수입니다.
```

2는 파랑과 빨강, 두 종류라는 뜻입니다. 집합은 리스트처럼 첫 위치를 고를 수 없습니다. 표식이 붙은 순서가 필요하면 colors에서 읽으세요.

### 두 값의 순서를 고정하기

튜플은 쉼표로 값을 묶고 보통 소괄호로 감쌉니다. 좌우 채널처럼 위치별 뜻이 정해졌고 칸을 바꾸지 않을 때 쓸 수 있습니다.

```python
channels = ("왼쪽", "오른쪽")
print(channels[1])  # 리스트처럼 0부터 위치를 셉니다.
```

오른쪽이 두 번째 채널입니다. channels[1] = "가운데"를 새 셀에서 실행하면 TypeError가 납니다. 튜플은 만든 뒤 칸에 새 값을 대입할 수 없습니다.

### 이름으로 칸을 찾기

딕셔너리는 중괄호 안에 키와 값을 짝지어 담습니다. 녹음마다 표식의 위치가 달라도 "색상"이라는 키로 값을 찾아 바꿀 수 있습니다.

```python
clip = {"번호": "07", "색상": "빨강"}
clip["색상"] = "초록"
print(clip["번호"], clip["색상"])  # print에 두 값을 넘기면 사이가 한 칸 떨어집니다.
```

번호 07은 그대로이고 색상만 초록이 됐습니다. 열쇠에 해당하는 키의 철자가 달라 clip["색"]을 읽으면 KeyError가 납니다. 키 이름을 먼저 확인하세요.

## 측정값에서 양의 짝수만 고르기

검사할 값이 한 줄에 모여 있습니다. 양수인지, 2로 나누어떨어지는지를 각각 판단한 다음 두 조건을 동시에 만족하는 값만 수집합니다. 같은 값이 여러 번 나오면 기록된 횟수대로 남겨야 합니다.

### 나눈 나머지로 짝수 확인하기

%는 왼쪽 수를 오른쪽 수로 나눈 나머지를 구합니다. 0도 짝수입니다. 그렇다면 방금 정한 두 조건을 모두 만족할까요?

```python
value = 0
print(value % 2 == 0)  # 2로 나눈 나머지가 0인지 봅니다.
print(value > 0)
```

짝수 조건은 True이고 양수 조건은 False입니다. and로 두 비교식을 이으면 둘 다 참일 때만 전체가 True가 됩니다. value를 -2로 바꿔도 두 번째 비교 결과를 먼저 예상할 수 있습니다.

### 조건에 따라 실행할 줄 고르기

if 다음에 조건을 적고 끝에 콜론을 붙입니다. 들여쓴 줄은 조건이 참일 때 실행합니다. else 아래에는 조건이 거짓일 때 할 일을 둡니다.

```python
value = 3
if value > 0 and value % 2 == 0:
    print("남김")
else:
    print("제외")  # 양수라도 홀수면 이 줄로 옵니다.
```

3은 2로 나누면 나머지 1이므로 제외됩니다. 조건을 읽을 때 3 > 0의 결과만 확인하면 홀수까지 남기게 됩니다.

### 여러 값에서 조건에 맞는 값 모으기

for는 리스트에서 값을 하나씩 꺼냅니다. in 앞의 이름은 매번 이번 차례의 값을 가리킵니다. 어떤 값도 통과하지 않을 수 있으므로 모아 둘 리스트는 반복을 시작하기 전에 비워 둡니다.

```python
values = [-2, 0, 3, 4]
positive = []
for value in values:
    if value > 0:
        positive.append(value)  # 두 번 들여썼으므로 조건을 통과한 차례에만 실행합니다.
print(positive)
```

3과 4가 순서대로 남습니다. 마지막 print는 왼쪽 끝에서 시작하므로 반복이 모두 끝난 뒤 한 번 실행합니다. positive = []를 반복 안으로 옮기면 앞서 모은 값이 사라집니다.

### 조건을 넣은 리스트 컴프리헨션 읽기

리스트 컴프리헨션은 새 리스트에 넣을 값, 값을 꺼낼 곳, 통과 조건을 대괄호 안에 모아 쓰는 문법입니다. 앞 셀을 짧게 쓰고 같은 순서로 남는지 봅시다.

```python
values = [-2, 0, 3, 4]
positive = [value for value in values if value > 0]
print(positive)  # for로 꺼내고 if로 거른 뒤 맨 앞의 값을 담습니다.
```

결과는 다시 [3, 4]입니다. 이번에는 왼쪽의 value를 value * 10으로 바꿔 보세요. 통과할 대상을 고르는 조건은 그대로이고, 담기는 값만 바뀝니다.

## 여러 번 불러도 기록이 섞이지 않는 함수

매번 같은 계산을 복사하면 단위를 고칠 때 여러 셀을 고쳐야 합니다. 계산에 이름을 붙여 여러 입력에 쓰고, 요청한 경우에만 이전 기록에 덧붙이도록 만들어 봅니다.

### 함수의 입력과 반환값

def 뒤에는 함수 이름과 입력을 받을 이름을 적습니다. 콜론 아래 들여쓴 부분은 함수를 호출할 때 실행됩니다. return은 계산값을 호출한 곳에 돌려줍니다.

```python
def to_seconds(minutes, seconds=0):
    # 초를 생략한 호출은 0초로 계산합니다.
    total = minutes * 60 + seconds
    return total

length = to_seconds(2, seconds=3)
print(length)
```

첫 번째 값 2는 순서에 따라 minutes에 들어갑니다. seconds=3은 이름을 지정한 키워드 인자입니다. to_seconds(2)로 호출하면 seconds의 기본값 0을 써서 120이 됩니다. 함수 안에서 만든 total은 지역 변수입니다. 함수 밖에서 계산 결과를 쓰려면 이 예제의 length처럼 반환값을 받아 두세요.

### 고칠 수 있는 입력 오류 처리하기

"2분"을 int()로 읽다가 만난 ValueError를 이번에는 처리합니다. try에 시도할 연산을 두고 except에서 그 오류가 발생했을 때 할 일을 정합니다.

```python
raw = "2분"
try:
    minutes = int(raw)
except ValueError:
    print("분 칸에 숫자만 입력하세요.")  # 어려운 오류 메시지 대신 고치는 방법을 보여 줍니다.
else:
    print(minutes * 60)
```

정수로 읽을 수 없는 자리에서 except로 이동했습니다. raw를 "2"로 바꾸면 오류 없이 try가 끝나므로 else가 실행되어 120이 됩니다. 잘못된 값을 0분으로 기록하면 실제로 0분인 기록과 구별할 수 없습니다.

### 기본값에 만든 리스트가 남는 이유

```python
def keep_label(label, labels=[]):
    labels.append(label)  # 호출할 때마다 이 목록을 바꿉니다.
    return labels

print(keep_label("시작"))
print(keep_label("종료"))
```

### 값을 받지 않았을 때만 준비하기

```python
def prepare_labels(labels=None):
    if labels is None:
        labels = []  # 호출자가 아무것도 주지 않은 이번 호출에서 새로 만듭니다.
    return labels

saved = ["리허설"]
print(prepare_labels(saved))
```

['리허설']은 호출자가 건넨 목록에 이미 들어 있었습니다. prepare_labels()의 결과는 []입니다. is None으로 검사하면 호출자가 명시적으로 준 빈 리스트도 그대로 쓸 수 있습니다.

## 두 녹음기의 누적 시간을 따로 관리하기

지금까지는 기록할 값을 매번 함수에 건넸습니다. 녹음기마다 시간이 얼마나 쌓였는지 기억하면서, 한 대만 골라 시간을 더하거나 0으로 되돌리고 싶습니다.

### 표준 모듈에서 함수 가져오기

모듈은 관련 기능을 이름으로 묶어 둔 것입니다. 표준 모듈은 파이썬과 함께 제공됩니다. math에서 ceil을 가져오면 초 단위 녹음을 분 단위 예약 시간으로 올림할 수 있습니다. /는 나눗셈입니다.

```python
from math import ceil

seconds = 61
print(ceil(seconds / 60))  # 1분을 조금 넘겨서 2분 자리를 예약합니다.
```

61초는 1분보다 길어서 2분이 필요합니다. 60초로 바꾸면 딱 1분입니다. from math import ceil에서 가져온 이름은 그 셀을 실행한 뒤 다음 셀에서도 쓸 수 있습니다.

### 인스턴스 하나의 값 바꾸기

클래스는 같은 종류의 값을 만들고 다루는 방법을 정의합니다. 클래스 이름 뒤에 괄호를 붙여 만든 개별 객체를 인스턴스라고 합니다. __init__은 새 인스턴스의 시작값을 넣는 메서드입니다. self는 지금 그 메서드를 사용하는 인스턴스를 가리킵니다.

```python
class Recorder:
    def __init__(self, name):
        self.name = name  # 이름을 인스턴스 자신의 속성에 보관합니다.

    def rename(self, name):
        self.name = name

left = Recorder("왼쪽")
right = Recorder("오른쪽")
left.rename("무대")
print(left.name, right.name)
```

무대 오른쪽이 출력됩니다. left.rename("무대")는 self에 left를 넘겨 메서드를 실행합니다. 같은 클래스로 만든 right의 name에는 손대지 않았습니다. 메서드 사이에서 계속 쓸 값은 self.name처럼 인스턴스의 속성에 둡니다.
