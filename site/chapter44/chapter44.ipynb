# 44장 · 평가 문항과 정답 근거 매핑

정답 근거는 평가 질문의 답을 뒷받침하는 문단입니다. 문서에 답이 없는 문항도 평가에 포함해야 임의의 답을 걸러낼 수 있습니다.

가상 운영 안내의 세 문단에 문항별 기대 근거를 연결합니다. 평일 수업 시각과 주말 일정처럼 표현은 가깝지만 답변 가능 여부가 다른 질문을 평가 자료에 포함합니다.

실행을 확인한 커널은 Python 3.12.15입니다.

Python 3 커널을 선택하고 위에서부터 실행하세요.


## 문단 ID와 기대 질문

Q01 “과제는 어디에 내나요?”는 p2에서 답할 수 있습니다. “과제함”만 들어 있으면 충분한지, 제출 뒤 파일 이름과 등록 시간을 확인하라는 점까지 필요한지 먼저 합의합니다. Q03의 업로드 오류는 p3에 연결합니다. 문단 번호와 실제 쪽수는 구별하세요. 44장 원자료에는 쪽수 필드가 없습니다.


### 셀 01 · 문항 ID와 정답 문단 매핑

질문 자료에 적힌 첫 기대 ID로 원문을 고릅니다.

Q01에서 ‘어디에’라는 질문의 필수 답은 과제함입니다. ‘제출 뒤 파일 이름과 등록 시간을 확인하세요’는 추가 안내로 기록하세요. 팀에서 절차 전체를 필수로 채점하려면 질문도 ‘어디에 내고 무엇을 확인하나요?’로 바꿉니다.


```python
paragraphs = [{'paragraph_id': 'p1',
  'text': '정규 수업은 평일 오전 9시에 시작합니다. 일정 변경은 과정 게시판에서 확인합니다.'},
 {'paragraph_id': 'p2',
  'text': '과제는 학습 관리 페이지의 과제함에 제출합니다. 제출 뒤 파일 이름과 등록 시간을 확인합니다.'},
 {'paragraph_id': 'p3',
  'text': '제출 오류가 발생하면 오류 화면과 발생 시간을 기록하고 담당 강사에게 문의합니다. 비밀키가 보이는 화면은 공유하지 '
          '않습니다.'}]
question = {'id': 'Q01',
 'split': 'development',
 'question': '과제는 어디에 내나요?',
 'answerable': True,
 'expected_paragraph_ids': ['p2'],
 'expected_points': ['과제함', '파일 이름과 등록 시간 확인']}
by_paragraph = {row['paragraph_id']: row['text'] for row in paragraphs}
print(by_paragraph[question['expected_paragraph_ids'][0]])
```

제출 위치와 제출 후 확인 절차가 한 문단에 함께 나옵니다. 검색기가 p3의 “제출 오류”를 골랐다면 단어는 비슷해도 질문의 답은 빠집니다.

정답 근거에는 제출 위치와 제출 후 확인할 내용이 함께 있습니다. 답변이 ‘과제함’만 말한다면 위치를 묻는 질문에는 충분할 수 있지만, 등록 여부까지 확인하려는 질문에는 부족합니다. 기대 답안을 작성할 때 질문이 요구하는 행동까지 문항별로 기록합니다.

### 셀 02 · 무응답 문항의 평가 근거

제공 평가 문항의 기대 조건과 p1 원문을 차례로 읽습니다.

#### 평일 안내문으로 주말 질문을 평가할 때

## 정답 근거를 달 수 없는 문항을 평가 묶음에서 찾기

자료 담당이 문단을 재편집하며 p3을 삭제했다고 가정합니다. 질문 표만 보면 여전히 답변 가능으로 표시된 Q03을 어떻게 발견할까요? 오늘 인계할 파일의 두 목록을 서로 대조합니다.

### 셀 03 · 평가표에서 원문으로 이어지는 참조 검사

p9는 잘못 입력한 문단 번호입니다. Q04는 의도적으로 정답 문단을 비워 둡니다. 아래 검사가 어느 문항을 보고하는지 보세요.

```python
items = {"Q01": ["p2"], "Q03": ["p9"], "Q04": []}

def unknown_evidence(questions, documents):
    return {qid: [key for key in refs if key not in documents]
            for qid, refs in questions.items() if any(key not in documents for key in refs)}

print(unknown_evidence(items, by_paragraph))
```

Q03의 p9는 course-guide-v1에서 열 수 없는 참조입니다. Q04의 빈 목록은 고장 난 링크가 아닙니다. 주말 운영을 뒷받침할 원문이 없다는 기대값입니다.

### 셀 04 · 원문을 읽고 근거 한 곳 바로잡기

오류 대응 절차가 적힌 문단을 원문에서 찾은 뒤 Q03에 그 ID를 넣습니다. 참조 검사와 실제 내용 확인을 차례로 실행하세요.

```python
fixed_items = {**items, "Q03": ["p3"]}
print(unknown_evidence(fixed_items, by_paragraph))
print(by_paragraph[fixed_items["Q03"][0]])
```

빈 사전은 모든 참조가 열린다는 뜻입니다. 두 번째 출력에서 “담당 강사에게 문의”까지 읽어야 p3이 질문의 정답 근거인지 판단할 수 있습니다.
