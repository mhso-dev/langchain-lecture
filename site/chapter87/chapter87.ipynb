## 질문·기대 행동·종료 조건

SC-01　교대 담당자가 “장비 A의 점검 전 확인 사항은?”이라고 묻습니다. 답에서 ‘전원 표시등’을 읽고 RULE-A 2.1절을 엽니다.

SC-02　“점검 전 확인 사항은?”이라고만 물으면 어느 장비인지 다시 묻습니다. SC-03　“장비 A를 다음 달에 교체하기로 확정했나요?”에는 가상 규정으로 일정을 확인할 수 없으니 담당자 확인을 안내합니다. SC-04　공개 그룹의 질문에는 관리용 문서가 보이지 않아야 합니다. 사용자 발화를 만들 때는 실제로 들은 말과 팀이 지어낸 질문을 구분해 남기세요.


### 셀 01 · 추가 질문이 필요한 조건

미리 고정한 개발 질문 Q02에서 첫 화면이 어떤 상태인지 읽어 보세요.

Q02의 화면에는 ‘장비 A인가요, 장비 B인가요?’라고 묻습니다. 개발 기록에는 need_clarification과 빈 출처 목록을 남깁니다. 팀 주제로 바꿀 때는 사용자가 답할 수 있는 두 선택지를 질문에 넣으세요.


````python
case = {'question_id': 'Q02',
 'scenario_id': 'SC-02',
 'question': '점검 전 확인 사항은?',
 'trusted_fixture_group': 'public',
 'expected_status': 'need_clarification',
 'expected_source_ids': [],
 'criterion_ids': ['CR-03'],
 'expected_answer_facts': ['장비 A 또는 장비 B']}
print(case["question"])
print(case["expected_status"], case["expected_source_ids"])
````

장비를 지정하지 않은 질문에는 need_clarification과 빈 목록이 붙습니다. 화면에는 A인지 B인지 묻는 문장을 두고, 답을 만들기 전에 장비명이 돌아오기를 기다립니다.

‘점검 전 확인 사항은?’에는 장비 구분이 없습니다. 규정은 있지만 어느 규정을 적용할지 정하지 못한 상태라서 추가 질문이 필요합니다. 공개 자료에 답 자체가 없는 질문과 구분하면 사용자가 정보를 더 주었을 때 처리를 이어 갈 수 있습니다.


#### 정보가 빠진 질문과 자료에 없는 질문

<mark>대상을 특정하지 못한 경우에는 정보를 더 받으면 답할 수 있습니다.</mark> 팀의 카드에서 ‘어느 장비인가요?’에 답하면 진행할 수 있는 질문과, 장비명을 받아도 규정에 없는 질문을 구별하세요. 두 경우의 종료 문구가 달라집니다.

## 한 문장이 빠진 대화로 되묻기와 종료를 구분하기

86장의 시연 의뢰에 동료가 “어떤 장비를 말하는지 안 적고 물어보면요?”라고 되묻습니다. 이때 추가 질문 한 번으로 복구할 수 있는 상황과 새 자료를 구해야 하는 상황을 대화로 나눕니다.

### 화면에 실제로 쓸 대화 초안

<table><thead><tr><th scope="col">첫 사용자 발화</th><th scope="col">서비스의 첫 응답</th><th scope="col">사용자의 다음 발화와 끝맺음</th></tr></thead><tbody><tr><td>점검 전 무엇을 보죠?</td><td>장비 A와 B 중 어느 장비인가요?</td><td>“B요.” 장비 B 점검 안내와 RULE-B 2.2절을 표시한다.</td></tr><tr><td>장비 A를 다음 달 교체하나요?</td><td>주어진 규정에서는 일정을 확인할 수 없습니다. 담당자에게 확인해 주세요.</td><td>같은 장비명을 다시 물어보지 않고 끝낸다.</td></tr></tbody></table>

첫 줄의 “B요”는 직전에 무엇을 물었는지 기억해야 해석됩니다. 첫 시연에 대화 기억을 넣지 않는 팀은 “장비 B의 점검 전 확인 사항은?”처럼 완성된 후속 질문을 입력받기로 명시하세요.

### 진행자와 기록자의 역할

진행자는 “저 규정으로 답할 수 있나요?”처럼 답을 암시하지 않습니다. 기록자는 참가자가 기입한 장비명, 서비스가 되물은 문장, 최종적으로 열린 절을 순서대로 적습니다. 표의 대화는 관측 전 시안입니다.

### 팀 자료에 붙여 보기

문의 양식에서 자주 비는 필드 하나를 고르세요. 그 필드를 보충한 뒤 같은 자료로 답을 낼 수 있는 질문과, 필드를 다 줘도 답할 수 없는 질문을 각각 만들어 동료에게 읽어 줍니다.

<mark>되물음이 해결할 수 있는 빈칸이 무엇인지 먼저 정합니다.</mark> Q02의 빈칸은 장비명입니다. Q03에는 장비 A가 이미 있고, 없는 것은 교체 일정을 확인할 문서입니다.
