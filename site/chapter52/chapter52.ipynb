Q03의 세 상태는 발표 문장을 구성하기 위해 옮겨 적은 기록입니다. 46장에서 후보가 p3을 찾았는데도 p2를 인용한 값을 보고, 50장에서 수정된 답의 ‘담당 강사에게 문의’와 p3 근거를 대조한 뒤 발표하세요.
