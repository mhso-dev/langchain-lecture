# 첫 모델 호출 전에 · 공지의 날짜와 장소 꺼내기

동아리 공지 C1에는 날짜와 장소가 함께 적혀 있습니다. 아래에서는 사람이 글을 읽고 두 값을 입력해 두었습니다. 날짜를 바꾸거나 장소만 꺼내려면 두 칸을 어떻게 구별할까요?

## 이름으로 값 찾기

딕셔너리는 이름(키)과 값의 짝을 보관합니다. 중괄호 안에서 따옴표로 적은 이름 오른쪽에 콜론을 쓰고 값을 놓습니다. 대괄호 안에 그 이름을 넣으면 짝지은 값을 하나 꺼낼 수 있습니다.

```python
notice = {"날짜": "5월 21일 금요일", "장소": "가온관 201호"}
print(notice["날짜"])  # 날짜라는 이름을 가진 칸을 읽습니다.
```

출력에는 날짜만 있습니다. 장소도 같은 공지에 들어 있지만 지금 읽은 칸은 날짜입니다. 키를 잘못 적었을 때 나타나는 KeyError는 그 이름의 칸을 찾지 못했다는 뜻입니다.

## 실습 · 바뀐 장소만 반영하기

공지 C2를 보면 모임 장소만 가온관 305호로 바뀌었습니다. notice["장소"]에 새 값을 대입하고 날짜와 장소를 차례로 출력해 보세요. 날짜는 C1과 같아야 합니다.

```python
notice = {"날짜": "5월 21일 금요일", "장소": "가온관 201호"}
# 이 아래에서 장소를 고치고 두 값을 각각 출력하세요.
```
