### 셀 01 · 과업 성공률의 분모

자료에 없는 실제 수강생 통계를 만들지 않도록 제공된 가상 네 시도만 사용합니다. 시작한 뒤 포기한 행은 분모에 남깁니다.


```python
attempts = [{'id': 'SYN-01', 'outcome': 'success'},
 {'id': 'SYN-02', 'outcome': 'failed'},
 {'id': 'SYN-03', 'outcome': 'aborted'},
 {'id': 'SYN-04', 'outcome': 'not_started'}]
started = [row for row in attempts if row["outcome"] != "not_started"]
successes = sum(row["outcome"] == "success" for row in started)
print(f"가상 과업 {len(started)}회 시작, {successes}회 성공")
```

시작 3회, 성공 1회입니다. 시작 전인 SYN-04만 뺐습니다. 가상 기록으로 과업 시도 횟수를 센 결과입니다. 참여자 수나 실제 사용자 성공률로 바꿔 읽으면 안 됩니다.

성공률의 분모는 시작한 과업 세 건입니다. 중도에 막힌 시도를 빼면 사용자가 겪은 실패가 지표에서 사라집니다. 세 건은 가상 관찰 자료이므로 실제 사용자를 만나기 전에는 실사용 성공률을 미측정으로 남깁니다.
