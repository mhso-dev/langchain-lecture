### 셀 01 · 로컬 조회 함수의 근거 범위

저장된 회귀 결과에서 Q01 한 건이 어떤 출처로 끝났는지 읽습니다.

이 팀의 결과물은 ‘장비 예약 API 연동 보류 메모’입니다. 다시 검토할 질문은 ‘이 장비를 실제로 예약할 수 있나요?’입니다. 현재 범위인 점검 안내에서 예약까지 맡기로 결정했을 때 권한과 운영 책임을 검토합니다.


```python
record = {'question_id': 'Q01', 'status': 'answered', 'source_ids': ['RULE-A']}
print(record["question_id"], record["status"], record["source_ids"])
```

Q01에서 answered와 RULE-A가 나옵니다. 교육용 로컬 경로로 이 한 건을 처리했다는 근거입니다. 실제 예약이나 현장 장비 정보에 접근한 기록은 없습니다.

Q01이 RULE-A로 답해졌다는 결과는 로컬 규정 조회를 검증합니다. 날짜에 따라 달라지는 공개 일정까지 확인했다는 근거는 없습니다. 외부 일정을 기능 범위에 넣으려면 읽을 수 있는 자료, 갱신 기준, 접근 실패 시 표시할 답을 따로 정합니다.


### 질문을 도구의 입력과 출력으로 나눠 보기
