## 수정 입력·출력으로 기여 범위 명시

**공백을 넣으면 어느 규정이 나올까요?** ‘장비A’, ‘장비B’, ‘장비AB’로 같은 점검 질문을 했다고 생각해 보세요. 수정 전에는 세 질문 모두 장비를 다시 지정하라는 답이 나왔습니다.

‘장비AB’에서 A 뒤에는 B가 붙어 있습니다. 두 글자를 그대로 둔 덕분에 다른 장비의 규정을 잘못 고르지 않았습니다.


### 셀 01 · 수정 범위의 입출력 근거

## 팀 사례에 적용하기 · 시연 근거와 후속 확인을 인수표로 묶기

위 예제에 첫 셀에서 읽어 둔 팀 기록을 넣어 봅니다.

```python
package = {"demo": incoming["scenes"], "local_regression": incoming["regression"],
           "deployment": incoming["deployment"]["status"], "peer_run": incoming["readme"]["peer_run"],
           "restore_missing": incoming["readme"]["missing_requests"]}
new = package
print(len(package["demo"]), package["deployment"], package["peer_run"])
```

실제 배포와 동료 재현은 각 기록이 확인됐을 때만 완료로 바꿉니다. 인수자는 시연 장면의 질문 하나를 그대로 실행한 뒤 자신의 관찰 결과를 덧붙일 수 있습니다.

```python
output_file = project_root / "team-project/final/chapter116.json"
output_file.parent.mkdir(parents=True, exist_ok=True)
with output_file.open("a+", encoding="utf-8") as handle:
    handle.seek(0)
    previous_text = handle.read()
    if previous_text and json.loads(previous_text) != new:
        raise FileExistsError("같은 장의 다른 결과가 있습니다. 기존 JSON을 다른 이름으로 보관한 뒤 다시 실행하세요.")
    if not previous_text:
        json.dump(new, handle, ensure_ascii=False, indent=2)
        handle.write("\n")
print(output_file.relative_to(project_root).as_posix())
```

팀 최종 인계 자료로 보관합니다. 입력을 바꿔 다시 실행할 때는 기존 결과 파일의 이름을 바꿔 보관하세요.
