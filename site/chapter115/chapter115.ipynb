## 재현 문서에서 정상·보류 시연 순서 만들기

114장의 재현 질문과 원문 위치를 큐시트 첫 장면에 넣습니다. 다음 장면은 근거 없는 질문이므로 이전 출처가 화면에서 사라지는지가 관찰 지점입니다.

이 프로젝트의 노트북은 한 폴더에서 열어 주세요. 첫 셀은 `team-project/final/chapter114.json`을 먼저 찾습니다. 팀 파일이 없으면 [이 장의 시작 예시](chapter115-start.json)를 노트북 옆에 내려받으세요. 출력 첫 줄에서 어느 파일을 읽었는지 확인할 수 있습니다.

````python
from pathlib import Path
import json
import re

here = Path.cwd()
project_root = here.parent if re.fullmatch(r"chapter[0-9]+", here.name) else here
team_file = project_root / "team-project/final/chapter114.json"
sample_file = here / "chapter115-start.json"
input_file = team_file if team_file.is_file() else sample_file
if not input_file.is_file():
    raise FileNotFoundError("이 장의 시작 예시를 내려받아 노트북 옆에 두세요.")
incoming = json.loads(input_file.read_text(encoding="utf-8"))
print("팀 기록" if input_file == team_file else "교재 예시", input_file.name)
print(incoming['readme'])
````

## 질문·답변·근거의 시연 순서

“점검 담당자는 교대할 때 장비 앞에서 무엇을 확인할지 원문을 다시 찾습니다. 저희는 답 옆에 규정의 절과 개정일을 붙였습니다. 장비 A를 물으면 ‘점검 전 전원 표시등을 확인합니다’라는 문장과 규정 A 2.1절이 나옵니다. 처음에는 장비 이름만 일치해도 답해서, 다음 달 교체 여부를 물었을 때까지 점검 규정을 보여 줬습니다. 지금 보실 로컬 예시는 그 질문에 담당자 확인을 안내합니다. 가상 규정으로 동작을 확인했고 실제 사용자가 더 빨리 찾는지는 다음 관찰에서 확인하겠습니다.”


### 셀 01 · 시연 답변과 근거 대응

발표자가 읽어 줄 답과 화면에서 가리킬 출처 위치를 표시합니다.


````python
rule = {'id': 'RULE-A',
 'equipment': '장비 A',
 'text': '점검 전 전원 표시등을 확인합니다.',
 'source': '가상 장비실 규정 A',
 'page': 1,
 'section': '2.1',
 'revised_on': '2026-11-01',
 'version': 'rules-v1',
 'groups': ['public', 'admin']}
print(rule["text"])
print(f'{rule["source"]} {rule["section"]}절 · 개정일 {rule["revised_on"]}')
````

원고의 전원 표시등과 화면의 2.1절을 대조할 수 있습니다. 원문과 화면 위치가 다르면 시연 전에 근거 카드부터 바로잡습니다.

발표 화면에 답이 뜬 뒤 실제 규정의 2.1절을 열어 같은 문장을 보여 줄 수 있어야 합니다. 이어서 근거가 없는 질문을 넣었을 때는 출처가 비는지 확인합니다. 두 장면을 연달아 시연하면 정상 답변과 답변 유보를 같은 기준으로 설명할 수 있습니다.


## 데모 실패 시 대체 자료와 전환 조건

### 팀 참고 예시 · 실패 사례 시연

## 팀 사례에 적용하기 · 재현 문서에서 정상·보류 시연 순서 만들기

위 예제에 첫 셀에서 읽어 둔 팀 기록을 넣어 봅니다.

````python
guide = incoming["readme"]
scenes = [{"question": guide["question"], "expected": guide["answer"], "locations": guide["source"]},
          {"question": "장비 A의 교체 일정은?", "expected": "일정 근거를 확인할 수 없음", "locations": []}]
new = {**incoming, "scenes": scenes}
print([len(scene["locations"]) for scene in scenes])
````

두 장면을 같은 순서로 실행하면 이전 답의 출처가 남는 오류를 찾을 수 있습니다. 발표 영상이나 화면 캡처를 실제 실행과 구분해 설명하세요.

````python
output_file = project_root / "team-project/final/chapter115.json"
output_file.parent.mkdir(parents=True, exist_ok=True)
with output_file.open("a+", encoding="utf-8") as handle:
    handle.seek(0)
    previous_text = handle.read()
    if previous_text and json.loads(previous_text) != new:
        raise FileExistsError("같은 장의 다른 결과가 있습니다. 기존 JSON을 다른 이름으로 보관한 뒤 다시 실행하세요.")
    if not previous_text:
        json.dump(new, handle, ensure_ascii=False, indent=2)
        handle.write("\n")
print(output_file.relative_to(project_root).as_posix())
````
