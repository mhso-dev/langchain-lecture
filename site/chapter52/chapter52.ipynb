## 재현 명세에서 시연할 장면 고르기

이 프로젝트의 노트북은 한 폴더에서 열어 주세요. 첫 셀은 `team-project/midterm/chapter51.json`을 먼저 찾습니다. 팀 파일이 없으면 [이 장의 시작 예시](chapter52-start.json)를 노트북 옆에 내려받으세요. 출력 첫 줄에서 어느 파일을 읽었는지 확인할 수 있습니다.

```python
from pathlib import Path
import json
import re

here = Path.cwd()
project_root = here.parent if re.fullmatch(r"chapter[0-9]+", here.name) else here
team_file = project_root / "team-project/midterm/chapter51.json"
sample_file = here / "chapter52-start.json"
input_file = team_file if team_file.is_file() else sample_file
if not input_file.is_file():
    raise FileNotFoundError("이 장의 시작 예시를 내려받아 노트북 옆에 두세요.")
incoming = json.loads(input_file.read_text(encoding="utf-8"))
print("팀 기록" if input_file == team_file else "교재 예시", input_file.name)
print(incoming['manifest'])
```

## 팀 사례에 적용하기 · 재현 명세에서 시연할 장면 고르기

위 예제에 첫 셀에서 읽어 둔 팀 기록을 넣어 봅니다.

```python
scenes = [row for row in incoming["manifest"] if row["id"] in incoming["fixes"] or not row["expected"]]
new = {"demo": scenes, "manifest_sha256": incoming["sha256"], "peer_run": incoming["peer_run"],
       "evaluated_path": incoming["evaluated_path"], "baseline_decision": incoming["baseline_decision"],
       "follow_up": "팀원이 새 환경에서 같은 질문으로 서비스 실행"}
print([(row["id"], row["question"], row["citations"]) for row in scenes])
```

수정한 Q03과 원래 보류 대상인 Q02가 서로 다른 장면으로 남습니다. 운영 성능이나 실제 이용자 반응은 이 로컬 시연으로 확인할 수 없습니다.

```python
output_file = project_root / "team-project/midterm/chapter52.json"
output_file.parent.mkdir(parents=True, exist_ok=True)
with output_file.open("a+", encoding="utf-8") as handle:
    handle.seek(0)
    previous_text = handle.read()
    if previous_text and json.loads(previous_text) != new:
        raise FileExistsError("같은 장의 다른 결과가 있습니다. 기존 JSON을 다른 이름으로 보관한 뒤 다시 실행하세요.")
    if not previous_text:
        json.dump(new, handle, ensure_ascii=False, indent=2)
        handle.write("\n")
print(output_file.relative_to(project_root).as_posix())
```
