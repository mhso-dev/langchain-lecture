## 새 폴더에서 같은 입력으로 결과 파일 재현하기

동료가 Q03을 다시 실행할 때 무엇을 건네야 할까요? 공개 가능한 작은 입력 파일과 기대 결과를 다른 폴더에 옮긴 뒤, 답변 안에 남은 문단 ID를 다시 읽어봅니다.

### 셀 03 · 입력 파일을 복사한 뒤 다시 열어 검산하기

아래 두 임시 폴더는 수업 예시 안에서만 사용합니다. 읽어들인 문자열에서 실제 인용을 추출한 뒤 기준 문단 ID와 비교합니다.

```python
from pathlib import Path
from tempfile import TemporaryDirectory
import json, shutil, re

fixture = {"question_id": "Q03", "answer": expected_answer, "expected_id": "course-guide-v1:p3"}
with TemporaryDirectory() as first, TemporaryDirectory() as second:
    source_file = Path(first) / "q03.json"
    source_file.write_text(json.dumps(fixture, ensure_ascii=False), encoding="utf-8")
    received = Path(shutil.copy(source_file, second))
    saved = json.loads(received.read_text(encoding="utf-8"))
    citation = re.search(r"\[([^\]]+)\]$", saved["answer"]).group(1)
    print(saved["question_id"], citation, citation == saved["expected_id"])
```

True는 복사한 결과 파일의 인용이 기대 문단과 같았다는 뜻입니다. 이 코드는 저장된 응답을 다시 읽었으며 검색기를 새로 설치하거나 답을 다시 생성하지 않았습니다.

## 팀 사례에 적용하기 · 회귀 결과를 재현용 입력과 묶기

위 예제에 첫 셀에서 읽어 둔 팀 기록을 넣어 봅니다.

```python
import hashlib
manifest = [{"id": row["id"], "question": row["question"], "expected": row["expected"],
             "citations": row["fixed_ids"], "passed": row["regression_ok"]} for row in incoming["runs"]]
payload = json.dumps(manifest, ensure_ascii=False, sort_keys=True).encode("utf-8")
new = {"manifest": json.loads(payload), "sha256": hashlib.sha256(payload).hexdigest(),
       "fixes": incoming["fixes"], "peer_run": "미확인",
       "evaluated_path": incoming["evaluated_path"], "baseline_decision": incoming["decision"]}
print(new["manifest"] == manifest, len(manifest), new["peer_run"])
```

다시 읽은 JSON의 일치는 전달 중 값이 보존됐다는 뜻입니다. 다른 사람이 새 환경에서 서비스까지 실행했는지는 peer_run에 따로 기록합니다.

```python
output_file = project_root / "team-project/midterm/chapter51.json"
output_file.parent.mkdir(parents=True, exist_ok=True)
with output_file.open("a+", encoding="utf-8") as handle:
    handle.seek(0)
    previous_text = handle.read()
    if previous_text and json.loads(previous_text) != new:
        raise FileExistsError("같은 장의 다른 결과가 있습니다. 기존 JSON을 다른 이름으로 보관한 뒤 다시 실행하세요.")
    if not previous_text:
        json.dump(new, handle, ensure_ascii=False, indent=2)
        handle.write("\n")
print(output_file.relative_to(project_root).as_posix())
```
