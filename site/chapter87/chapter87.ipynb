## 첫 질문에 되묻기와 범위 밖 질문 더하기

이 프로젝트의 노트북은 한 폴더에서 열어 주세요. 첫 셀은 `team-project/final/chapter86.json`을 먼저 찾습니다. 팀 파일이 없으면 [이 장의 시작 예시](chapter87-start.json)를 노트북 옆에 내려받으세요. 출력 첫 줄에서 어느 파일을 읽었는지 확인할 수 있습니다.

````python
from pathlib import Path
import json
import re

here = Path.cwd()
project_root = here.parent if re.fullmatch(r"chapter[0-9]+", here.name) else here
team_file = project_root / "team-project/final/chapter86.json"
sample_file = here / "chapter87-start.json"
input_file = team_file if team_file.is_file() else sample_file
if not input_file.is_file():
    raise FileNotFoundError("이 장의 시작 예시를 내려받아 노트북 옆에 두세요.")
incoming = json.loads(input_file.read_text(encoding="utf-8"))
print("팀 기록" if input_file == team_file else "교재 예시", input_file.name)
print(incoming['scenario'])
````

### 셀 01 · 추가 질문이 필요한 조건

미리 고정한 개발 질문 Q02에서 첫 화면이 어떤 상태인지 읽어 보세요.

Q02의 화면에는 ‘장비 A인가요, 장비 B인가요?’라고 묻습니다. 개발 기록에는 need_clarification과 빈 출처 목록을 남깁니다. 팀 주제로 바꿀 때는 사용자가 답할 수 있는 두 선택지를 질문에 넣으세요.


````python
case = {'question_id': 'Q02',
 'scenario_id': 'SC-02',
 'question': '점검 전 확인 사항은?',
 'trusted_fixture_group': 'public',
 'expected_status': 'need_clarification',
 'expected_source_ids': [],
 'criterion_ids': ['CR-03'],
 'expected_answer_facts': ['장비 A 또는 장비 B']}
print(case["question"])
print(case["expected_status"], case["expected_source_ids"])
````

장비를 지정하지 않은 질문에는 need_clarification과 빈 목록이 붙습니다. 화면에는 A인지 B인지 묻는 문장을 두고, 답을 만들기 전에 장비명이 돌아오기를 기다립니다.

‘점검 전 확인 사항은?’에는 장비 구분이 없습니다. 규정은 있지만 어느 규정을 적용할지 정하지 못한 상태라서 추가 질문이 필요합니다. 공개 자료에 답 자체가 없는 질문과 구분하면 사용자가 정보를 더 주었을 때 처리를 이어 갈 수 있습니다.

## 사용자별 시나리오 카드

시나리오 카드 예시　상황 SC-03 / 담당자가 다음 달 교체 여부를 물음 / 사용할 수 있는 자료는 점검 규정 / 화면은 일정을 확인할 수 없다고 안내 / 사용자는 실제 담당자에게 확인. 아직 사용자를 만나지 않았다면 ‘가상 과업’으로 표시하고, 실제 관찰을 받았을 때 어느 부분이 바뀌었는지 적습니다.


### 팀 참고 예시 · 공개 범위의 시나리오


````python
group_cases = [{'question_id': 'Q05',
  'scenario_id': 'SC-04',
  'question': '관리자 전용 절차를 알려 주세요.',
  'trusted_fixture_group': 'public',
  'expected_status': 'no_evidence',
  'expected_source_ids': [],
  'criterion_ids': ['CR-05'],
  'expected_answer_facts': ['확인할 수 없']},
 {'question_id': 'Q07',
  'scenario_id': 'SC-04',
  'question': '관리자 전용 절차를 알려 주세요.',
  'trusted_fixture_group': 'admin',
  'expected_status': 'answered',
  'expected_source_ids': ['RULE-ADMIN'],
  'criterion_ids': ['CR-05', 'CR-02'],
  'expected_answer_facts': ['선임 검토 기록']}]
public_case = next(row for row in group_cases if row["trusted_fixture_group"] == "public")
print(public_case["expected_status"])
````

## 한 문장이 빠진 대화로 되묻기와 종료를 구분하기

86장의 시연 의뢰에 동료가 “어떤 장비를 말하는지 안 적고 물어보면요?”라고 되묻습니다. 이때 추가 질문 한 번으로 복구할 수 있는 상황과 새 자료를 구해야 하는 상황을 대화로 나눕니다.

### 화면에 실제로 쓸 대화 초안

<table><thead><tr><th scope="col">첫 사용자 발화</th><th scope="col">서비스의 첫 응답</th><th scope="col">사용자의 다음 발화와 끝맺음</th></tr></thead><tbody><tr><td>점검 전 무엇을 보죠?</td><td>장비 A와 B 중 어느 장비인가요?</td><td>“B요.” 장비 B 점검 안내와 RULE-B 2.2절을 표시한다.</td></tr><tr><td>장비 A를 다음 달 교체하나요?</td><td>주어진 규정에서는 일정을 확인할 수 없습니다. 담당자에게 확인해 주세요.</td><td>같은 장비명을 다시 물어보지 않고 끝낸다.</td></tr></tbody></table>

첫 줄의 “B요”는 직전에 무엇을 물었는지 기억해야 해석됩니다. 첫 시연에 대화 기억을 넣지 않는 팀은 “장비 B의 점검 전 확인 사항은?”처럼 완성된 후속 질문을 입력받기로 명시하세요.

### 진행자와 기록자의 역할

진행자는 “저 규정으로 답할 수 있나요?”처럼 답을 암시하지 않습니다. 기록자는 참가자가 기입한 장비명, 서비스가 되물은 문장, 최종적으로 열린 절을 순서대로 적습니다. 표의 대화는 관측 전 시안입니다.

### 팀 자료에 붙여 보기

문의 양식에서 자주 비는 필드 하나를 고르세요. 그 필드를 보충한 뒤 같은 자료로 답을 낼 수 있는 질문과, 필드를 다 줘도 답할 수 없는 질문을 각각 만들어 동료에게 읽어 줍니다.

<mark>되물음이 해결할 수 있는 빈칸이 무엇인지 먼저 정합니다.</mark> Q02의 빈칸은 장비명입니다. Q03에는 장비 A가 이미 있고, 없는 것은 교체 일정을 확인할 문서입니다.

## 팀 사례에 적용하기 · 첫 질문에 되묻기와 범위 밖 질문 더하기

위 예제에 첫 셀에서 읽어 둔 팀 기록을 넣어 봅니다.

````python
first_case = incoming["scenario"]
scenarios = [{"id": first_case["id"], "q": first_case["question"], "status": "answered",
              "source_id": first_case["source_id"]},
             {"id": "Q02", "q": case["question"], "status": case["expected_status"], "source_id": None},
             {"id": "Q03", "q": "장비 A를 다음 달에 교체하나요?", "status": "no_evidence", "source_id": None}]
new = {"scenarios": scenarios, "acceptance": incoming["acceptance"]}
print([(row["id"], row["status"]) for row in scenarios])
````

장비가 빠진 행은 추가 질문, 일정 자료가 없는 행은 근거 없음으로 끝납니다. 팀 기획에서 장비명이 달라졌다면 첫 질문의 source_id와 원문도 함께 바꿉니다.

````python
output_file = project_root / "team-project/final/chapter87.json"
output_file.parent.mkdir(parents=True, exist_ok=True)
with output_file.open("a+", encoding="utf-8") as handle:
    handle.seek(0)
    previous_text = handle.read()
    if previous_text and json.loads(previous_text) != new:
        raise FileExistsError("같은 장의 다른 결과가 있습니다. 기존 JSON을 다른 이름으로 보관한 뒤 다시 실행하세요.")
    if not previous_text:
        json.dump(new, handle, ensure_ascii=False, indent=2)
        handle.write("\n")
print(output_file.relative_to(project_root).as_posix())
````
