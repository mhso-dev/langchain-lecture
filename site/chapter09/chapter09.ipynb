### 셀 01 · 같은 입력으로 시작하기

가상 교육센터 원문에서 두 비용이 담긴 줄을 찾아 보세요.

```python
notice = """가상 교육센터 안내
온라인 설명회는 12월 5일 오후 3시에 열린다.
참가 신청은 12월 3일 오후 6시에 마감한다.
설명회 참가비는 무료다. 교육과정의 수강료는 아직 발표하지 않았다.
다시보기 제공 여부는 확정되지 않았다.
신청 주소는 추후 안내한다."""
v1 = "이 안내문 정리해줘."
print(next(line for line in notice.splitlines() if "수강료" in line))
```

### 셀 06 · 수강료가 발표된 새 원문 만들기

‘미발표’라는 고정 답을 외운 요청이라면 다음 입력에서 어떻게 될까요?

원문에서 수강료만 바꾸었습니다. 요청은 그대로 두고 새 원문으로 확인해야 무엇이 응답 차이를 만들었는지 알 수 있습니다.

```python
revised = notice.replace("교육과정의 수강료는 아직 발표하지 않았다.", "교육과정의 수강료는 5만 원이다.")
print(next(line for line in revised.splitlines() if "수강료" in line))
```

새 원문에서 수강료는 5만 원입니다. 지시는 ‘항상 미발표라고 적기’ 대신 ‘이번 자료가 밝힌 상태를 유지하기’여야 바뀐 정보도 전달합니다.
