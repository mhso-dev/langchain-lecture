### 셀 03 · 미실행 시나리오의 분모 제외

제공 기록에서 발췌한 두 행입니다. 실제 로컬 실행을 표시한 행만 골라 봅니다.

정상 질문 4행은 답과 근거를 확인하는 데 사용합니다. 진단 20행은 실패 위치를 찾는 기록이고 미채택 2행은 후보를 쓰지 않은 이유를 남깁니다. 세 묶음을 합친 26행을 모두 정상 정답 문항으로 세지 마세요.


```python
entries = [{"case_id": "E67-search-error", "execution_status": "executed_local", "failure_stage": "retrieval"},
           {"case_id": "E67-memory-unselected", "execution_status": "not_selected"}]
observed = [row for row in entries if row["execution_status"] == "executed_local"]
print(observed[0]["case_id"], observed[0]["failure_stage"])
```

### 셀 05 · 출처 ID 검사와 기대 문장 검사를 따로 실행하기

이 한 건은 원문의 표현을 그대로 인용하기로 한 회귀 시험입니다. 자유롭게 바꿔 쓴 답변 전체를 자동 채점하는 규칙은 아닙니다. 기존 candidate를 시험 입력에 넣으세요.

```python
case = {"question": "과제는 어디에 내나요?", "expected_id": source["id"],
        "expected_quote": "과제는 학습 관리 페이지의 과제함에 제출합니다."}

def check_quote(case, docs, answer):
    hits = {doc["id"] for doc in docs}
    if case["expected_id"] not in hits:
        return "retrieval"
    if f"[{case['expected_id']}]" not in answer:
        return "citation"
    if case["expected_quote"] not in answer:
        return "answer"
    return "passed"

print(check_quote(case, [source], candidate))
```

answer가 나옵니다. 검색 결과에 기대 문서가 있었고 답 안에 인용 ID도 있었지만, 개인 메신저라는 안내가 기대 원문과 달랐습니다. 앞선 검사를 통과한 뒤 실패한 곳이므로 먼저 살펴볼 코드는 답변을 만드는 부분입니다.
