## 응답의 출처로 알파 관찰 과업 정하기

102장 응답에서 원문 절을 꺼내 ‘도움 없이 근거 찾기’가 끝나는 지점을 정합니다. 실제 사용자 반응을 수집하기 전에는 결과를 예시 관찰로 표시합니다.

이 프로젝트의 노트북은 한 폴더에서 열어 주세요. 첫 셀은 `team-project/final/chapter102.json`을 먼저 찾습니다. 팀 파일이 없으면 [이 장의 시작 예시](chapter103-start.json)를 노트북 옆에 내려받으세요. 출력 첫 줄에서 어느 파일을 읽었는지 확인할 수 있습니다.

```python
from pathlib import Path
import json
import re

here = Path.cwd()
project_root = here.parent if re.fullmatch(r"chapter[0-9]+", here.name) else here
team_file = project_root / "team-project/final/chapter102.json"
sample_file = here / "chapter103-start.json"
input_file = team_file if team_file.is_file() else sample_file
if not input_file.is_file():
    raise FileNotFoundError("이 장의 시작 예시를 내려받아 노트북 옆에 두세요.")
incoming = json.loads(input_file.read_text(encoding="utf-8"))
print("팀 기록" if input_file == team_file else "교재 예시", input_file.name)
print(incoming['reply'])
```

두 번째 줄은 읽은 파일의 `reply` 값입니다.

## 출처 탐색의 관찰 사실과 해석

가상 알파 기록　과업은 “장비 A를 점검하기 전 확인할 사항과 근거 절을 찾아 주세요.” 참여자 가-01은 답의 전원 표시등을 읽은 뒤 화면을 멈췄습니다. 진행자가 출처가 보이는지 물은 뒤에야 2.1절을 골랐습니다. 관찰은 ‘힌트 뒤 절을 찾음’입니다. ‘글씨가 작아서’는 아직 확인하지 않은 해석입니다.


### 셀 01 · 관찰 사실과 추정 구분

이 카드는 형식을 보여 주기 위해 채운 가상 기록입니다. 학생 팀에서는 실제 과업을 수행한 사람이 한 행동을 적습니다.

다음 시안에서는 출처의 위치만 답 바로 아래로 옮겨 보세요. 링크 문구와 글자 크기는 그대로 두고, 다음 참가자가 힌트 없이 2.1절을 찾는지 기록합니다. 그래야 위치를 바꾼 뒤 무엇을 관찰했는지 설명할 수 있습니다.


```python
observation = {"종류": "가상 관찰 예시", "과업": "Q01의 근거 절 찾기",
               "행동": "답을 읽고 멈췄다가 힌트 뒤에 2.1절 선택",
               "가설": "답과 출처 사이 간격 때문에 출처를 놓쳤을 수 있음"}
print(observation["행동"])
```

행동만 꺼냈으므로 사용자가 왜 멈췄는지는 출력에 없습니다. “어디를 보셨나요?”처럼 이유를 유도하지 않는 후속 질문을 따로 준비할 수 있습니다.

관찰한 사실은 사용자가 답을 읽다가 멈췄고 힌트 뒤에 2.1절을 골랐다는 것입니다. 왜 멈췄는지는 아직 모릅니다. 표기를 수정한 다음 다른 과업에서 도움 없이 개정일을 찾는지 관찰하면 설명 방식에 대한 가설을 시험할 수 있습니다.


#### 사용성 메모에서 해석을 꺼내는 순서

<mark>‘힌트 뒤에 2.1절을 선택했다’는 관찰이고 왜 멈췄는지는 가설입니다.</mark> 팀원이 본 행동을 먼저 적고 안내 문구 하나만 고쳐 같은 과업을 맡기세요. 다음 관찰에서 도움 없이 개정일을 찾았는지가 수정 판단의 근거가 됩니다.

### 같은 관찰로 설명할 수 있는 두 가지 가설

가상 참가자가 힌트 뒤 2.1절을 골랐다는 행동은 관찰할 수 있습니다. 출처가 멀어서 놓쳤을 수도 있고, 원문까지 찾아야 한다는 과업을 다르게 이해했을 수도 있습니다. 문구와 위치 중 한 가지를 바꾸고 같은 과업에서 힌트가 필요했는지 기록해 보세요.


## 관찰에 근거한 단일 화면 수정

## 팀 사례에 적용하기 · 응답의 출처로 알파 관찰 과업 정하기

위 예제에 첫 셀에서 읽어 둔 팀 기록을 넣어 봅니다.

```python
source = next(iter(incoming["reply"]["sources"]), None)
target = {"id": source["id"], "section": source["section"]} if source else None
new = {**incoming, "alpha": {"task": "답변의 원문 절 열기", "target": target, "observed": None}}
print(new["alpha"])
```

완료 지점이 특정 문서와 절이므로 ‘화면이 좋았다’ 같은 감상과 구별할 수 있습니다. 실제 참여자가 없으면 observed 값을 비워 두세요.

```python
output_file = project_root / "team-project/final/chapter103.json"
output_file.parent.mkdir(parents=True, exist_ok=True)
with output_file.open("a+", encoding="utf-8") as handle:
    handle.seek(0)
    previous_text = handle.read()
    if previous_text and json.loads(previous_text) != new:
        raise FileExistsError("같은 장의 다른 결과가 있습니다. 기존 JSON을 다른 이름으로 보관한 뒤 다시 실행하세요.")
    if not previous_text:
        json.dump(new, handle, ensure_ascii=False, indent=2)
        handle.write("\n")
print(output_file.relative_to(project_root).as_posix())
```
