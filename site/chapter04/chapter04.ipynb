## __init__과 self로 인스턴스 상태 저장

클래스는 새 기록에 어떤 값을 담고 어떤 일을 시킬지 정합니다. StudyRecord를 두 번 부르면 같은 규칙을 따르는 별개의 객체가 생깁니다.

### 셀 01 · __init__과 인스턴스 생성

생성 시 받은 값과 기본 완료 표시가 어떻게 저장되는지 보세요.


```python
class StudyRecord:
    def __init__(self, title: str, minutes: int):
        self.title = title
        self.minutes = minutes
        self.done = False

    def mark_done(self) -> None:
        self.done = True

    def to_dict(self) -> dict:
        return {"title": self.title, "minutes": self.minutes, "done": self.done}

first = StudyRecord("Python 기초", 30)
print(first.to_dict())
```

done에 False가 들어 있습니다. self는 메서드를 호출한 바로 그 객체입니다. to_dict는 지금 그 객체 안에 든 값들을 사전으로 꺼냅니다.


#### 서로 다른 두 수업의 완료 상태

<mark>self는 이번에 메서드를 호출한 객체를 가리킵니다.</mark> 같은 클래스로 수업 두 개를 만들고 한쪽의 완료 상태만 바꿔 보세요. 두 객체의 제목과 상태를 각각 읽으면 어느 데이터가 함께 묶이는지 확인할 수 있습니다.

### 셀 02 · 인스턴스별 속성 분리

첫 기록만 완료로 바꾼 뒤 방금 만든 둘째 기록을 확인합니다.


```python
second = StudyRecord("Git 연습", 20)
first.mark_done()
print(second.done)
```

False가 나옵니다. first.mark_done()의 self는 first를 가리키므로 second.done은 처음 값 그대로입니다.

second를 만들 때 StudyRecord를 다시 호출한 줄을 보세요. 그 줄을 second = first로 바꾸면 두 이름이 한 기록을 가리킵니다. 이때 first.mark_done()을 실행하면 second.done으로 읽어도 True입니다. 별도의 기록이 필요할 때 객체를 한 번 더 만드는 이유입니다.

### 셀 04 · 생성 함수와 int() 변환

공백이 많은 제목과 30이라는 문자열을 넘겨 봅니다.


```python
def create_record(title: str, minutes_text: str) -> StudyRecord:
    title = " ".join(title.split())
    if not title:
        raise ValueError("제목이 비어 있습니다.")
    minutes = int(minutes_text)
    if minutes < 0 or minutes > 480:
        raise ValueError("분은 0~480 사이여야 합니다.")
    return StudyRecord(title, minutes)

entry = create_record("  Python   기초 ", "30")
print(entry.to_dict())
```

제목에는 공백 한 칸이 남고 minutes에는 정수 30이 들어 있습니다. 원본 패키지에서 service는 models의 StudyRecord를 가져다 씁니다. 노트북에서는 위 셀에서 정의한 같은 클래스를 씁니다.

### 셀 07 · 모듈 간 import와 실행

제공 셀이 임시 studylog 폴더에 두 모듈을 만듭니다. models에서 클래스를 정의한 줄과 service에서 그 클래스를 불러오는 줄을 나란히 읽으세요.

models.py에서 기록의 구조를 정의하고 service.py에서 그 기록을 사용합니다. from .models의 점은 같은 패키지 안에서 models를 찾는다는 뜻입니다. 먼저 이 두 파일의 내용만 읽고 아래 제공 셀을 통째로 실행하세요.


```python
import subprocess
import sys
from pathlib import Path
from tempfile import TemporaryDirectory

models_text = """class StudyRecord:
    def __init__(self, title):
        self.title = title
"""
service_text = """from .models import StudyRecord

def make_record(title):
    return StudyRecord(title.strip())
"""
with TemporaryDirectory() as folder:
    package = Path(folder) / "studylog"
    package.mkdir()
    (package / "__init__.py").write_text("", encoding="utf-8")
    (package / "models.py").write_text(models_text, encoding="utf-8")
    (package / "service.py").write_text(service_text, encoding="utf-8")
    command = 'from studylog.service import make_record; print(make_record("  Git 연습  ").title)'
    result = subprocess.run([sys.executable, "-c", command], cwd=folder,
                            check=True, capture_output=True, text=True)
    print(result.stdout.strip())
```

‘Git 연습’이 나옵니다. service의 from .models에서 점 하나는 같은 패키지를 뜻합니다. 방금 실행한 작은 패키지는 제목 정리만 옮긴 예입니다. 전체 학습 기록 패키지에는 앞에서 확인한 시간 검사와 완료 표시도 있습니다.
